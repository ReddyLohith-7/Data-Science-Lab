```python
# AIM:
# To analyze customer purchasing behavior and divide customers
# into different groups (clusters) using the K-Means clustering
# algorithm.
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
url = "https://raw.githubusercontent.com/tirthajyoti/Machine-Learning-with-Python/master/Datasets/Mall_Customers.csv"
df = pd.read_csv(url)
```

```python
print("=" * 70)
print("FIRST FIVE RECORDS")
print("=" * 70)

print(df.head())


print("\n")
print("=" * 70)
print("DATASET INFORMATION")
print("=" * 70)

df.info()


print("\n")
print("=" * 70)
print("DATASET SHAPE")
print("=" * 70)

print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])


print("\n")
print("=" * 70)
print("COLUMN NAMES")
print("=" * 70)

print(df.columns.tolist())
```

```python
print("\n")
print("=" * 70)
print("MISSING VALUES")
print("=" * 70)

print(df.isnull().sum())
```

```python
print("\n")
print("=" * 70)
print("DUPLICATE RECORDS")
print("=" * 70)

print("Number of duplicate rows:", df.duplicated().sum())
```

```python
X = df[
    [
        "Annual Income (k$)",
        "Spending Score (1-100)"
    ]
]


print("\n")
print("=" * 70)
print("SELECTED FEATURES")
print("=" * 70)

print(X.head())
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)


print("\n")
print("=" * 70)
print("STANDARDIZED FEATURES")
print("=" * 70)

print(X_scaled[:5])
```

```python
inertia = []

for k in range(1, 11):

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    model.fit(X_scaled)

    inertia.append(model.inertia_)


# Plot Elbow Curve

plt.figure(figsize=(8, 5))

plt.plot(
    range(1, 11),
    inertia,
    marker="o"
)

plt.title("Elbow Method for Optimal Number of Clusters")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")

plt.grid(True)

plt.tight_layout()

plt.savefig("elbow_method.png")

plt.show()
```

```python
kmeans = KMeans(
    n_clusters=5,
    random_state=42,
    n_init=10
)
```

```python
df["Cluster"] = kmeans.fit_predict(X_scaled)
```

```python
print("\n")
print("=" * 70)
print("CUSTOMER CLUSTERS")
print("=" * 70)

print(
    df[
        [
            "CustomerID",
            "Annual Income (k$)",
            "Spending Score (1-100)",
            "Cluster"
        ]
    ].head(20)
)
```

```python
cluster_centers = scaler.inverse_transform(
    kmeans.cluster_centers_
)

centers_df = pd.DataFrame(
    cluster_centers,
    columns=[
        "Annual Income (k$)",
        "Spending Score (1-100)"
    ]
)

centers_df.index.name = "Cluster"


print("\n")
print("=" * 70)
print("CLUSTER CENTERS")
print("=" * 70)

print(centers_df)
```

```python
print("\n")
print("=" * 70)
print("CUSTOMER COUNT PER CLUSTER")
print("=" * 70)

print(df["Cluster"].value_counts().sort_index())
```

```python
summary = df.groupby("Cluster")[
    [
        "Annual Income (k$)",
        "Spending Score (1-100)"
    ]
].mean()

summary["Customer Count"] = df["Cluster"].value_counts()

summary = summary.sort_index()


print("\n")
print("=" * 70)
print("CUSTOMER SEGMENT SUMMARY")
print("=" * 70)

print(summary)
```

```python
plt.figure(figsize=(10, 7))

sns.scatterplot(
    data=df,
    x="Annual Income (k$)",
    y="Spending Score (1-100)",
    hue="Cluster",
    palette="tab10",
    s=100
)


# Plot cluster centers

plt.scatter(
    cluster_centers[:, 0],
    cluster_centers[:, 1],
    marker="X",
    s=250,
    c="black",
    label="Cluster Centers"
)


plt.title(
    "Customer Personality Analysis using K-Means",
    fontsize=15
)

plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.savefig(
    "customer_segments_kmeans.png",
    dpi=300
)

plt.show()
```

```python
df.to_csv(
    "Mall_Customers_Clustered.csv",
    index=False
)
```
